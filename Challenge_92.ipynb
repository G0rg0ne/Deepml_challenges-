```python
import numpy as np
from sklearn.linear_model import LinearRegression
import math
```

```python
PI = 3.14159

model = LinearRegression()

def substract_fluctuations(consumption_data):
    denoised_data = [consumption_data[i] - (10 * np.sin(2 * PI * i / 10)) for i in range(len(consumption_data))]
    return denoised_data

def power_grid_forecast(consumption_data):
    denoised_data = substract_fluctuations(consumption_data)
    model.fit(np.arange(1,len(denoised_data)+1).reshape(-1, 1), denoised_data)
    day_number = 15
    day_fifthen = model.predict(np.array([day_number]).reshape(-1, 1)) + (10 * np.sin(2 * PI * day_number / 10))
    # 5) Round, then add a 5% safety margin (rounded up).
    day_fifthen = math.ceil(day_fifthen * 1.05)
    # Return the final integer.
    return int(day_fifthen)
```

```python
daily_consumption = np.array([150, 165, 185, 195, 210, 225, 240, 260, 275, 290]) 
```

```python
power_grid_forecast(daily_consumption)
```
