```python
import numpy as np
```

```python

def impute_missing_data(data: np.ndarray, strategy: str = 'mean') -> np.ndarray:
    """
    Impute missing values in a 2D array using the specified strategy.
    
    Args:
        data: 2D numpy array with missing values represented as np.nan
        strategy: Imputation strategy - 'mean', 'median', or 'mode'
        
    Returns:
        2D numpy array with missing values imputed
    """
    n,m = data.shape
    for column in range(m):
        column_data = data[:,column]
        if np.isnan(column_data).any():
            if strategy == 'mean':
                mean_value = np.mean(column_data[~np.isnan(column_data)])
                column_data[np.isnan(column_data)] = mean_value
                data[:,column] = column_data
            elif strategy == 'median':
                median_value = np.median(column_data[~np.isnan(column_data)])
                column_data[np.isnan(column_data)] = median_value
                data[:,column] = column_data
            elif strategy == 'mode':
                vals = column_data[~np.isnan(column_data)] 
                mode_value = mode_value = np.unique(vals, return_counts=True)[0][np.unique(vals, return_counts=True)[1].argmax()]
                column_data[np.isnan(column_data)] = mode_value
                data[:,column] = column_data
    return data
```

```python
data = [[1.0, 2.0], [3.0, np.nan], [5.0, 6.0]]
strategy = 'mean'
```

```python
impute_missing_data(data,strategy)
```
